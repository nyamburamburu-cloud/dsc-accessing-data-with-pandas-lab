Next, we'd like the information on all the games played in **Group 3** for the **1950** World Cup.

Hint: You can combine conditions like this:

`df[(condition1) & (condition2)]`  -> Returns rows where both conditions are true

```python
# Your code here
group_3_games_1950 = df[(df['Year'] == 1950) & (df['Stage'] == 'Group 3')]
print(group_3_games_1950)
```

Throughout the entire history of the World Cup as recorded in this dataset, **how many home games were played by the Netherlands**?

(Remember that you can use the `len()` built-in function to find the number of rows in a DataFrame.)

```python
# Your code here
# Filter the DataFrame for home games played by the Netherlands
netherlands_home_games = df[df['Home Team Name'] == 'Netherlands']

# Count the number of home games
home_games_count = len(netherlands_home_games)
print(home_games_count)
```

**How many games were played by the Netherlands in total**?

```python
# Your code here
# Filter the DataFrame for games played by the Netherlands as either home or away team
netherlands_games = df[(df['Home Team Name'] == 'Netherlands') | (df['Away Team Name'] == 'Netherlands')]

# Count the total number of games
total_games_count = len(netherlands_games)
print(total_games_count)
```

Next, let's try and figure out **how many games the USA played in the 2014 World Cup**.

```python
# Your code here 
# Step 1: Filter the DataFrame for games played in the 2014 World Cup
usa_games_2014 = df[df['Year'] == 2014]

# Step 2: Count the number of games where the USA was either the home or away team
usa_game_count = len(usa_games_2014[(usa_games_2014['Home Team Name'] == 'USA') | (usa_games_2014['Away Team Name'] == 'USA')])
print("USA games in 2014 World Cup:", usa_game_count)
```

Now, let's try to find out **how many countries participated in the 1986 World Cup**.

Hint 1: As a first step, create a new dataset that only contains games in that year.

Hint 2: Make sure you don't end up with duplicate country names. Consider using `set()` or `.unique()`.

```python
# Your code here
# Step 1: Filter the DataFrame for games played in the 1986 World Cup
world_cup_1986 = df[df['Year'] == 1986]

# Step 2: Create a set of unique country names from both home and away teams
unique_countries = set(world_cup_1986['Home Team Name']).union(set(world_cup_1986['Away Team Name']))

# Step 3: Count the number of unique countries
number_of_countries = len(unique_countries)
print("Number of countries in the 1986 World Cup:", number_of_countries)
```

## Changing Values and Creating New Columns

In World Cup history, **how many matches had 5 goals or more in total**? Create a column `"Total Goals"` to answer this question.

```python
# Your code here
# Step 1: Create a new column "Total Goals" by summing the goals of both teams
df['Total Goals'] = df['Home Team Goals'] + df['Away Team Goals']

# Step 2: Filter the DataFrame for matches with 5 or more total goals
matches_with_5_goals = df[df['Total Goals'] >= 5]

# Step 3: Count the number of matches with 5 or more goals
number_of_matches_with_5_goals = len(matches_with_5_goals)
print("Number of matches with 5 or more goals:", number_of_matches_with_5_goals)
```

Now **create a new column `"Half-time Goals"`** in `df` that includes both home and away values.

```python
# Your code here
# Step 1: Check the actual column names
print(df.columns)

df['Half-time Goals'] = df['Home Half-time Goals'] + df['Away Half-time Goals']
```

Run the code below. You'll notice that for Korea, there are records for both North-Korea (Korea DPR) and South-Korea (Korea Republic). 

```python
# Run this cell without changes

# Display all records containing the string 'Korea'
df.loc[df['Home Team Name'].str.contains('Korea'), 'Home Team Name']
```

Imagine that, for some reason, we simply want Korea listed as one entry, so we want to replace every "Home Team Name" and "Away Team Name" entry that contains "Korea" to simply "Korea". In the same way, we want to change the columns "Home Team Initials" and "Away Team Initials" to NSK (North & South Korea) instead of "KOR" and "PRK". 

```python
# Update the 'Home Team Name' and 'Home Team Initials' columns 
# Step 1: Update 'Home Team Name' to "Korea" where it contains "Korea"
df['Home Team Name'] = df['Home Team Name'].replace(to_replace=r'.*Korea.*', value='Korea', regex=True)

# Step 2: Update 'Home Team Initials' to "NSK" where it is "KOR" or "PRK"
df['Home Team Initials'] = df['Home Team Initials'].replace({'KOR': 'NSK', 'PRK': 'NSK'})

# Step 3: Repeat for 'Away Team Name' and 'Away Team Initials'
df['Away Team Name'] = df['Away Team Name'].replace(to_replace=r'.*Korea.*', value='Korea', regex=True)
df['Away Team Initials'] = df['Away Team Initials'].replace({'KOR': 'NSK', 'PRK': 'NSK'})
```

Make sure to verify your answer!

```python
# Check the updated columns
# Step 1: Check unique values in 'Home Team Name' and 'Home Team Initials'
unique_home_team_names = df['Home Team Name'].unique()
unique_home_team_initials = df['Home Team Initials'].unique()

# Step 2: Print the unique values to verify the updates
print("Unique Home Team Names:", unique_home_team_names)
print("Unique Home Team Initials:", unique_home_team_initials)

# Step 3: Repeat for 'Away Team Name' and 'Away Team Initials'
unique_away_team_names = df['Away Team Name'].unique()
unique_away_team_initials = df['Away Team Initials'].unique()

print("Unique Away Team Names:", unique_away_team_names)
print("Unique Away Team Initials:", unique_away_team_initials)
```
